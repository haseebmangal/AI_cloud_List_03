```python
#list methods investigation
#sort(), reverse(), index(), count()
#sort()
letters = ["c", "g", "h", "a", "t", "b", "o"]
letters.sort()
print(letters)
```

```python
# reverse()
letters.reverse()
print(letters)
```

```python
# index
letters_1 = letters.index("t")
print(letters_1)
```

```python
# len(), sum(), min(), max()
#len()
scores = [29, 87, 56, 43, 23, 17, 67]
len_score = len(scores)
print(len_score)
```

```python
# sum()
sum_score = sum(scores)
print(sum_score)
```

```python
# min()
minimum = min(scores)
print(minimum)
```

```python
# max()
maximum = max(scores)
print(maximum)
```

```python
average_score = sum_score/len_score
print(average_score)
```

```python
# Topic no 29: set
# craete set
class_topping = {"pizza", "sauce", "cheese", "onions", "lemonade", "sauce", "onions", "cheese"}
print(class_topping)
```

```python
# add an element into set
# we will use add() keyword
add_class_topping = class_topping.add("margarita")
print(class_topping)
```

```python
# update() method is used to add multiple elements at once
update_class_topping = class_topping.update("burger", "coffee")
print(class_topping)
```
