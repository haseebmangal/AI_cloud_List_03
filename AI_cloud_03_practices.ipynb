```python
# exercise 01
# create a list named movie
movies = ["007", "titanic", "Godzilla", "avengers", "Mercy for none", "007", "spiderman", "batman", "spiderman", "avengers"]

# printing the whole list
print(movies)
```

```python
# print the first and last movie
Ist_movie = movies[0]
print(Ist_movie)
last_movie = movies[-1]
print(last_movie)
```

```python
# f string
print(f"first movie : {movies[0]} \
and last movie : {movies[-1]}"\
)
```

```python
# append a new movie
added_movie = movies.append("hero")
print(movies)
```

```python
# exercise 02
# remove a movie
remove_movie = movies.remove("hero")
print(movies)
```

```python
# length of movies
len_movies = len(movies)
print(len_movies)
```

```python
# sort the list alphabatically with .sort()
sort_movies = movies.sort()
print(movies)
```

```python
# check the index with .count()
count_movies = movies.count("spiderman")
print(count_movies)
```

```python
# tuple
# create a tuple
my_location = ("peshawar", 25000, "saddar", "house no 29", 234567.879)
print(my_location)
```

```python
# now printing each value with indexing
# indexing each element location
index_01 = my_location.index("peshawar")
print(index_01)
```

```python
index_02 = my_location.index("saddar")
print(index_02)
```

```python
# finding the error
my_location[0] = 10
# TypeError: 'tuple' object does not support item assignment
```

```python
# dictionary practices
me = {
    "name" : "haseeb",
    "age" : 29,
    "course" : " python"
}
print(me)
```

```python
# now updating an existing element
me["age"] = 30
print(me)
```

```python
# add a new key value pair
me["Hobbies"] = ["Boxing", "swimming", "MMA"]
print(me)
```
